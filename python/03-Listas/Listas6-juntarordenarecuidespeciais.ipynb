# Juntar Listas, Ordenar e Cuidados Especiais

### 2 formas:

- lista1.extend(lista2)
- lista_nova = lista1 + lista2

Obs: o Método .append não junta listas, mas adiciona um valor no final da lista

```python
produtos = ['apple tv', 'mac', 'iphone x', 'iphone 11', 'IPad', 'apple watch', 'mac book', 'airpods']
novos_produtos = ['iphone 12', 'ioculos']

#Incluindo novos registros a lista existente
produtos.extend(novos_produtos)
print(produtos)

#Incluindo novos registros, mas criando uma nova 
produtos2 = produtos + novos_produtos
print(produtos2)

#Não é possivel utilizar append para listas, pois você na verdade, você vai adicionar uma lista dentro de outra.
```

### Cuidado:

- [1] + [2] não é a mesma coisa que 1 + 2, então cuidado sempre com o formato dos valores na hora de fazer operações.

```python
vendas = [1000, 1500, 15000, 20000, 270, 900, 100, 1200]
vendas_iphonex = [15000]
vendas_iphone11 = [20000]

# print([1]+[2]) é diferente de print(1+2)
totaliphone = vendas[2] + vendas [3]
print(totaliphone)

totaliphones = vendas_iphone11 + vendas_iphonex
print(totaliphones)
totaliphones2 = vendas_iphone11[0] + vendas_iphonex[0]
print(totaliphones2)
```

### Ordenar listas

lista.sort()

```python
#Ordena sempre das palavras com maiusculo p/ minusculo e AZ
produtos.sort()
print(produtos)
```

```python
vendas.sort(reverse=True)
print(vendas)
```
