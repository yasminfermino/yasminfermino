# Algumas Funções Básicas de Lista

## Tamanho da Lista

tamanho = len(lista)

```python
produtos = ['apple tv', 'mac', 'iphone x', 'IPad', 'apple watch', 'mac book', 'airpods']
```

- Quantos produtos temos a venda?

```python
tamanho = len(produtos)
print(f'Temos {tamanho} a venda.')
```

## Maior e Menor Valor

maior = max(lista)

menor = min(lista)

```python
vendas = [1000, 1500, 15000, 270, 900, 100, 1200]
```

- Qual o item mais vendido?
- Qual o item menos vendido?

```python
maisvend = max(vendas)
menosvend = min(vendas)
```

```python
produtomaisvend = vendas.index(maisvend)
produtomenosvend = vendas.index(menosvend)

produtomaisvend = produtos[produtomaisvend]
produtomenosvend = produtos[produtomenosvend]

print(f'O produto mais vendido é {produtomaisvend} com total de {maisvend} e o menos vendido foi {produtomenosvend} com o total de {menosvend}.')
```
