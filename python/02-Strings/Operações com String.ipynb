# Operações com String

str -> transforma número em string<br>
in -> verifica se um texto está contido dentro do outro<br>
operador + -> concatenar string<br>
format e {} -> substitui valores<br>
%s -> substitui textos<br>
%d -> substitui números decimais<br>

Vamos deixar uma cartilha para download

```python
faturamento = 1000
custo = 500
lucro = faturamento - custo
```

Uso do str() e do concatenar com +

```python
print ('O faturamento da loja foi de: ' + str(faturamento)) #problema desse forma é que se você precisar usar em varios momentos ou se precisar usar varias variaveis pode dar bastante trabalho
```

Uso do Format

```python
print('O faturamento foi de: {0}. O custo for de: {1} e o faturamento que foi {0}, superou ano passado'.format(faturamento, custo))
# {} - caso seja necessario indicar entre os colchetes qual o indice da informmação a ser usada.
```

Uso do %s e %d

```python
print ('O faturamento foi de: %d ' % (faturamento))

#Funcionam de forma parecida com o format (%d - substitui por um numero e %s - substitui por uma string)
```

Uso do in (mais exercícios práticos nas próximas aulas)

```python
print('@' in 'lira@gmail.com')
print('@' in 'lira.gmail.com')

#verifica se tem info passada dentro do texto seguinte, ou da lista, etc
```
