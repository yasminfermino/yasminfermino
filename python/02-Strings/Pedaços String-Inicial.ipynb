### Forma Básica

```python
precos = "Jan: 25, Fev: 27, Mar: 29"

precojan = precos[5:7]
print(precojan)

precofev = precos[14:16]
print(precofev)

precomar = precos[-2:]
print(precomar)

precofev2 = precos[-11:-9]
print(precofev2)

#Da direita p/ esquerda
```

### Posição Inicial e Final com Step

```python
codigo = "1.2.3.4,5,1,2.3.4,7.9"
pedacocod = codigo[::2] #temos o step, o passo com o qual ela anda na lista/ string

print(pedacocod)
```

```python
codigo = "1.2.3.4,5,1,2.3.4,7.9"
pedacocod = codigo[10:2:-2] #está da esq para direita

print(pedacocod)
```
