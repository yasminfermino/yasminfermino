#### 3. Faça um programa que leia e valide as seguintes informações (e para cada uma delas, continue pedindo a informação até o usuário inserir corretamente):
##### Nome: maior que 3 caracteres;
##### Idade: entre 0 e 150;
##### Salário: maior que zero;
##### Sexo: 'f' ou 'm';
##### Estado Civil: 's', 'c', 'v', 'd';

````python

nome = input('Seu nome:')
while True:
    if len(nome) < 3:
        nome = input('Por favor, informe novamente seu nome, ele não pode ter menos que 3 caracteres:')
    else:
        break

idade = int(input('Qual sua idade:'))
while not (0 < idade <= 150):
    idade = int(input('Por favor informe uma idade valida, entre 0 e 150 anos:'))
    
salario = float(input('Qual seu salario:'))
while not salario > 0:
    salario = float(input('Por favor informar um salario maior que zero:')) 

sexo = input('Informe qual seu genero [f] ou [m]:').lower()
while not (sexo == 'f' or sexo == 'm'):
    sexo = input('Por favor informar o sexo com a letra [f] para feminino ou [m] para masculino:').lower()

estcivil = input('Qual seu estado civil? [s] solteiro(a), [c] casado(a), [v] viuvo(a) ou [d] divorsiado(a)').lower()

while not estcivil in ['s','c','v','d']:
    estcivil = input('Por favor informar o estado civil com o uso das seguintes letras: [s] solteiro(a), [c] casado(a), [v] viuvo(a) ou [d] divorsiado(a)').lower()

print([nome, idade, salario, sexo, estcivil])
````

Supondo que a população de um país A seja da ordem de 80000 habitantes com uma taxa anual de crescimento de 3% e que a população de B seja 200000 habitantes com uma taxa de crescimento de 1.5%. Faça um programa que calcule e escreva o número de anos necessários para que a população do país A ultrapasse ou iguale a população do país B, mantidas as taxas de crescimento.

````python
ano = 0
cida = 80000
cidb = 200000
taxaa = 0.03
taxab = 0.015

while cida < cidb:
    ano += 1
    cida = (cida * taxaa)+ cida
    cidb = (cidb * taxab)+ cidb

print(f'Será necessario {ano} anos, para que A ultrapasse B em quantidade de pessoas. Cidade A com {cida:.0f} pessoas e Cidade B com {cidb:.0f} pessoas.')
````

#### 6. Faça um programa que peça para o usuário inserir o faturamento dos últimos 5 meses (individualmente) e informe o maior faturamento

````python
faturamentos = []
i = 1

while i < 5:
    faturamento = float(input(f'Insira o faturamento do mês:'))
    faturamentos.append(faturamento)

    i += 1

fattotal = sum(faturamentos)
media = fattotal/len(faturamentos)

print(f'O faturamento total dos primeiros cinco meses foi de {fattotal:.0f} e sua média de faturamento foi de {media:.0f}.')
````

````python
faturamentos = {}
i = 1

while i <= 5:
    mes = input(f'Insira o nome do mês {i}: ')
    faturamento = float(input(f'Insira o faturamento do mês de {mes}: '))
    faturamentos[mes] = faturamento
    print (f'O {mes} teve o faturamento de {faturamento} Reais')
    i += 1

print(faturamentos)
fattotal = sum(faturamentos.values())
media = fattotal / len(faturamentos)

print(f'O faturamento total dos primeiros cinco meses foi de {fattotal:.0f} Reais e sua média de faturamento foi de {media:.0f} Reais.')
````

#### 9. Numa eleição existem três candidatos. Faça um programa que peça o número total de eleitores. Peça para cada eleitor votar e ao final mostrar o número de votos de cada candidato.

````python
from collections import Counter
candidatos = {'Yasmin':'20', 'Gabriel':'13', 'Giovanni':'24'}

votos = []

while True:
    voto = input('Vote em seu candidato "Número de dois dig":')
    if not voto:
        break
    else:
        for  itens in candidatos.values():
            if  voto in itens:
                votos.append(voto)
            else:
                pass

contador = Counter(votos)


for chave, item in candidatos.items():
    for chave2, item2 in contador.items():
        if item == chave2:
            print(f'Candidato {chave}, recebeu {item2} votos;')
        else:
            pass

#print(contador)
````

#### 10. Faça um programa que calcule o valor total investido por um colecionador em sua coleção de CDs e o valor médio gasto em cada um deles. O usuário deverá informar a quantidade de CDs e o valor para em cada um.

````python
# Entrada 

compras = {}
valor_gasto = 0
n = 0

while True:
    nome_ep = input('Nome do CD (caso deseja enterromper a inclusão de CDs aperte Enter)')
    if nome_ep:  # Verifica se o nome do CD foi fornecido
        valor_ep = float(input('Quanto pagou no CD:'))

        compras[nome_ep] = valor_ep # nome_ep é a chave e valor_ep é o valor associado
    else:
        break

for valor_ep in compras.values(): #Olhando valores do dicionario, precisamos colocar o nome do dict e '.values' para ele trazer somente valores gastos
    n += 1
    valor_gasto +=  valor_ep 

gasto_medio = valor_gasto/n

print(f'Valor total gasto com os CDs foi de {valor_gasto:.2f} reais e valor médio foi de {gasto_medio:.2f} reais')
````
