## 1. Criando um Registro de Hóspedes

Digamos que você está criando o sistema para registrar a chegada de hóspedes em um hotel. No hotel, os hóspedes podem ter quartos com 1, 2, 3 e 4 pessoas. Seu sistema deve conseguir:

1. Identificar quantas pessoas o hóspede que acabou de chegar vai ter no quarto (perguntando por meio de input)
2. De acordo com a quantidade de pessoas do hóspede, ele deve fazer um for para perguntar o cpf e o nome de cada pessoa, a fim de registrá-la no quarto (2 inputs para cada pessoa, 1 para o cpf e outro para o nome)
3. O seu programa então deve gerar uma lista com todas as pessoas que ficarão no quarto em que cada item dessa lista é o nome da pessoa e o cpf da pessoa, assim:

- Para simplificar, não vamos nos preocupar com possibilidades de "tentar colocar mais de 1 hóspede, digitar o cpf errado, etc. Nosso objetivo é treinar a criação de uma rotina de cadastro

```python
def format_cpf(cpf):
    """Formata um CPF no formato XXX.XXX.XXX-XX."""
    cpf = ''.join(filter(str.isdigit, cpf))  # Remove caracteres não numéricos
    
    if len(cpf) != 11:
        cpf = ('Digite novamente o CPF.')
        return cpf
    
    return f"{cpf[:3]}.{cpf[3:6]}.{cpf[6:9]}-{cpf[9:]}"

# Pede a quantidade de hóspedes
qtd_hospedes = int(input('Qual a quantidade de hóspedes? (1 a 4): '))

# Lista para armazenar as informações dos hóspedes
lista_final = []

# Loop para coletar informações dos hóspedes
for i in range(qtd_hospedes):
    nome = input('Inclua o nome do hóspede: ')
    cpf = input('Insira CPF do hóspede (Apenas números): ')
    cpf_formatado = format_cpf(cpf)
    
    # Verifica se o CPF é válido
    if cpf_formatado:
        lista_final.append({"nome": nome, "cpf": cpf_formatado})

# Exibe as informações dos hóspedes
print('Informações dos hóspedes:')
for hospede in lista_final:
    print(f"Nome: {hospede['nome']}, CPF: {hospede['cpf']}")
```

## 3. Comparação com Ano Anterior

Digamos que você está analisando as vendas de produtos de um ecommerce e quer identificar quais produtos tiveram no ano de 2020 mais vendas do que no ano de 2019, para reportar isso para a diretoria.

Sua resposta pode ser um print de cada produto, qual foi a venda de 2019, a venda de 2020 e o % de crescimento de 2020 para 2019.

Lembrando, para calcular o % de crescimento de um produto de um ano para o outro, podemos fazer: (vendas_produto2020/vendas_produto2019 - 1)

Dica: lembre do enumerate, ele pode facilitar seu "for"

```python
produtos = ['iphone', 'galaxy', 'ipad', 'tv', 'máquina de café', 'kindle', 'geladeira', 'adega', 'notebook dell', 'notebook hp', 'notebook asus', 'microsoft surface', 'webcam', 'caixa de som', 'microfone', 'câmera canon']
vendas2019 = [558147,712350,573823,405252,718654,531580,973139,892292,422760,154753,887061,438508,237467,489705,328311,591120]
vendas2020 = [951642,244295,26964,787604,867660,78830,710331,646016,694913,539704,324831,667179,295633,725316,644622,994303]
#seu código aqui


```

```python
for i, produto in enumerate(produtos):
    if vendas2019[i] < vendas2020[i]:
        aumento = ((vendas2020[i] - vendas2019[i])/vendas2020[i])*100
        print(f'O produto {produto} teve em vendas de 2019 foi de {vendas2019[i]} reais e em 2020 de {vendas2020[i]} reais. Isto representa um aumento de {aumento:.2f}%')
    else:
        pass

print ('Fim do programa')
```

```python
''' ':.2f': Isso é uma especificação de formatação que define como o valor de aumento será exibido na string. Vamos explicar cada parte:

':': Indica que estamos começando a especificação de formatação do valor da variável.

'.2f': Significa que queremos formatar o número com 2 casas decimais após a vírgula (ou ponto, dependendo da configuração regional).

'f': A letra "f" indica que estamos lidando com um número ponto flutuante (float). Ou seja, valores que podem ter decimais.

'.2': O número 2 após o ponto indica que queremos 2 casas decimais. Se o valor de aumento for, por exemplo, 12.345, ele será exibido como 12.35.'''
```
