#### 4.Em quais meses a média de temperatura foi maior do que a média nacional?

````python
meses = [
    'Janeiro',
    'Fevereiro',
    'Março',
    'Abril',
    'Maio',
    'Junho',
    'Julho',
    'Agosto',
    'Setembro',
    'Outubro',
    'Novembro',
    'Dezembro'
]

temperaturas = [30, 29, 28, 28, 25, 26, 20, 21, 19, 25, 27, 32]

media = sum(temperaturas)/12

print("A temperatura foi maior nós seguintes meses do ano:")

for i, temp in enumerate(temperaturas):
    if temp > media:
        print(f'{meses[i]}', end= "/")

````

````python
lista_salarios = [1000, 300, 500, 200, 1500, 3000, 3400, 5000, 7000, 2000, 600, 800, 250, 1500, 20000]
abonos = []
valormin = 0

for salario in lista_salarios:
    abono = 0.2*salario
    if abono < 100:
        abono = 100
        abonos.append(abono)
        valormin += 1
    else:
        abonos.append(abono)
    print(f'Salário: {salario} Reais - Abono: {abono} Reais')

print(f'Foram processados {len(lista_salarios)} colaboradores')
print(f'Total gastos com abonos: R${sum(abonos)}')
print(f'Valor mínimo pago a {valormin} colaboradores')
print(f'Maior valor de abono pago: R$ {max(abonos)}')
````
